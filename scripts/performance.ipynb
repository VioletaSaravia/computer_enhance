```python
import functools as ft

def cycles_per_byte(freq, throughput):
    return (throughput * 1024**3) / (freq * 1000**3)

e14_cpb = ft.partial(cycles_per_byte, 1.8)
t14_cpb = ft.partial(cycles_per_byte, 3.3)
```

```python
print(f"{e14_cpb(3.284):.2f} bytes/cycle")
print(f"{e14_cpb(4.040):.2f} bytes/cycle")
```
